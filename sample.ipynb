```python
class Bank:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance   # Private variable

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Invalid amount")

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")

    def get_balance(self):
        return self.__balance


# Object creation
b = Bank("Ram", 10000)

print("Account Holder:", b.name)
print("Initial Balance:", b.get_balance())

b.deposit(5000)
print("Balance after deposit:", b.get_balance())

b.withdraw(3000)
print("Balance after withdrawal:", b.get_balance())

b.withdraw(15000)   # Insufficient balance
print("Final Balance:", b.get_balance())
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.__available = True

    def display(self):
        print("Title:", self.title)
        print("Author:", self.author)

    def borrow(self):
        if self.__available:
            print("Book borrowed successfully")
            self.__available = False
        else:
            print("Book not available")

# Create object
b = Book("Python", "Abdul Kalam")

# Display book details
b.display()

# Borrow the book twice
b.borrow()
b.borrow()
```
